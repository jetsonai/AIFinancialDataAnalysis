## 환경 2. 로컬 AI(Ollama) 연결 확인

Ollama가 켜져 있으면 이 PC에 받아 둔 모델 목록이 나옵니다.

```python
import json, urllib.request
import llm_config

try:
    models = [m["name"] for m in json.load(urllib.request.urlopen("http://127.0.0.1:11434/api/tags", timeout=5))["models"]]
    print("Ollama 실행 중 · 받아 둔 모델:", models)
    print("수업용 모델", llm_config.OLLAMA_MODEL, "→", "있음 ✅" if llm_config.OLLAMA_MODEL in models else "없음 ❌ (명령 프롬프트에서 ollama pull qwen3:4b-instruct)")
except Exception as e:
    print("Ollama에 연결할 수 없습니다 ❌ → 작업 표시줄에서 Ollama를 실행했는지 확인하세요.", e)
```
