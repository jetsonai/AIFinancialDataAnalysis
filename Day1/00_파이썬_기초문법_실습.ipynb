## 24쪽. 조건으로 행 고르기

- `==` 같은가? → 결과는 True / False
- `&` = 그리고 (조건마다 괄호로 감쌉니다)
- 조건은 True/False 목록을 만들고 **True인 행만** 남깁니다.

```python
df["지역"] == "광주"          # 79개 회사마다 True / False
```

```python
df[df["지역"] == "광주"]      # True인 행만 → 광주 5개사
```

두 번째 조건 예시는 **고객 데이터**(`bank_customers.csv`)로 해 봅니다. 저축은행 표와 섞이지 않도록 이름을 `고객`으로 붙입니다.

```python
고객 = pd.read_csv("bank_customers.csv")

# 주택담보대출은 있고(housing == "yes") 개인대출은 없는(loan == "no") 고객
주담대만 = 고객[(고객["housing"] == "yes") & (고객["loan"] == "no")]
print("주택담보대출만 있는 고객:", len(주담대만), "명")      # 결과: 20763 명
```

> ⚠️ **`=`(넣기)와 `==`(같은가)를 구분합니다.**
> 아래 셀은 일부러 `==` 대신 `=`를 써서 **에러를 내 보는** 셀입니다. 실행하면 빨간 `SyntaxError`가 나오는 것이 정상입니다.
> 에러를 확인했으면 `=`를 `==`로 고쳐 다시 실행해 보세요.

```python
# [일부러 내는 에러] = 를 == 로 고치면 광주 5개사가 나옵니다
df[df["지역"] = "광주"]
```

## 25쪽. groupby와 함수 만들기

`groupby` = **엑셀 피벗 테이블**입니다. 오늘 가장 많이 쓰는 문법입니다.

**행 전체** → **지역별 바구니로 나누기** → **바구니마다 합계**

```python
df.groupby("지역")["총자산"].sum()          # 지역별 총자산 합계 (억원)
```

### 자주 쓰는 계산에 이름 붙이기 — `def`

같은 계산을 여러 번 해야 하면 **함수**로 만들어 이름을 붙여 둡니다. 함수 안쪽은 읽기만 하고, **쓰는 법**(`가입률표("기준")`)만 기억하면 됩니다.
먼저 21쪽의 단어장처럼 영문 직업명을 한글로 바꿔 둡니다(예제 02에서 자세히 배웁니다).

```python
직업명 = {
    "admin.": "사무직", "blue-collar": "생산·현장직", "entrepreneur": "사업가",
    "housemaid": "가사도우미", "management": "관리직", "retired": "은퇴자",
    "self-employed": "자영업", "services": "서비스직", "student": "학생",
    "technician": "기술직", "unemployed": "무직", "unknown": "미상",
}
고객["직업"] = 고객["job"].map(직업명)     # 단어장대로 바꿔 새 열 만들기
```

```python
def 가입률표(기준):
    표 = 고객.groupby(기준)["y"].agg(
        고객수="size",
        가입자수=lambda x: (x == "yes").sum())
    표["가입률(%)"] = (표["가입자수"] / 표["고객수"] * 100).round(1)
    return 표.sort_values("가입률(%)", ascending=False)

가입률표("직업")          # 결과: 학생 28.7% · 은퇴자 22.8% … 생산·현장직 7.3%
```

✏️ **직접 해 보기**: 기준만 바꾸면 다른 표가 바로 나옵니다. `가입률표("marital")`(결혼 여부)을 실행해 보세요.

```python
가입률표("marital")
```
